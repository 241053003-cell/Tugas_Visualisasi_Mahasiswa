```python
import pandas as pd
import numpy as np

np.random.seed(42)
n_samples = 1000

# 1. Kota (Kategorik Nominal) - 5 kota Indonesia
kota_list = ['Jakarta', 'Bandung', 'Surabaya', 'Medan', 'Yogyakarta']
prob_kota = [0.30, 0.20, 0.20, 0.15, 0.15]
kota = np.random.choice(kota_list, n_samples, p=prob_kota)

# 2. Tanggal (untuk analisis time series)
tanggal = pd.date_range(start='2026-01-01', periods=n_samples, freq='H')
# Kita acak urutannya agar lebih realistis
tanggal = np.random.choice(tanggal, n_samples, replace=False)
tanggal = pd.to_datetime(tanggal)

# 3. Suhu (°C) - Numerik Interval
# Setiap kota punya profil suhu berbeda
suhu_profile = {
    'Jakarta':    (31.0, 2.0),  # Panas, kota pesisir
    'Bandung':    (22.0, 3.0),  # Sejuk, dataran tinggi
    'Surabaya':   (33.0, 2.5),  # Sangat panas
    'Medan':      (30.0, 2.5),  # Panas, tropis
    'Yogyakarta': (28.0, 2.5),  # Sedang
}
suhu = np.array([np.random.normal(suhu_profile[k][0], suhu_profile[k][1]) for k in kota])
suhu = np.round(suhu, 1)

# 4. Kelembaban (%) - Numerik Rasio
# Berbanding terbalik dengan suhu (semakin panas, semakin kering biasanya)
kelembaban_base = 90 - (suhu - 22) * 2
kelembaban = kelembaban_base + np.random.normal(0, 5, n_samples)
kelembaban = np.clip(kelembaban, 30, 100)
kelembaban = np.round(kelembaban).astype(int)

# 5. Kondisi Cuaca (Kategorik Nominal)
kondisi_list = ['Cerah', 'Berawan', 'Hujan', 'Berangin']
# Probabilitas kondisi berdasarkan kelembaban
def tentukan_kondisi(kelembaban_val):
    if kelembaban_val > 85:
        return np.random.choice(['Hujan', 'Berawan'], p=[0.7, 0.3])
    elif kelembaban_val > 70:
        return np.random.choice(['Berawan', 'Cerah', 'Hujan'], p=[0.5, 0.3, 0.2])
    else:
        return np.random.choice(['Cerah', 'Berangin'], p=[0.7, 0.3])

kondisi = np.array([tentukan_kondisi(k) for k in kelembaban])

# 6. Membuat DataFrame
df = pd.DataFrame({
    'Tanggal': tanggal,
    'Kota': kota,
    'Suhu': suhu,
    'Kelembaban': kelembaban,
    'Kondisi': kondisi
})

# Urutkan berdasarkan tanggal
df = df.sort_values('Tanggal').reset_index(drop=True)

# Simpan
df.to_csv('data_cuaca.csv', index=False)
print("✅ Dataset 1000 baris cuaca berhasil dibuat!")
print(f"📅 Rentang tanggal: {df['Tanggal'].min()} s/d {df['Tanggal'].max()}")
print(f"🌍 Kota: {df['Kota'].unique().tolist()}")
df.head(10)
```
